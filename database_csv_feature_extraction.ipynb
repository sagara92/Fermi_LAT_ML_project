# 'blazar_catalog_v35.csv' reduction and feature extraction.

```python
import pandas as pd
import numpy as np
import ast
```

# Load the data file and define the features of interest

```python
# Open the blazar catalog file and extract the following features.
"""------Features used in iopscience.iop.org/article/10.3847/1538-4357/ab558b/meta paper--- 
"PL_Index","LP_Index","Pivot_Energy","PLEC_Flux_Density","LP_Flux_Density",
"PL_Flux_Density","Frac2_Variability","nuFnu_Band7","PLEC_Index","Flux_Band2",
"Flux_Band7","nuFnu_Band2","Frac_Variability","PLEC_Expfactor","Variability2_Index",
"Variability_Index","nuFnu_Band6","Flux_Band3","Npred","Flux_Band6","nuFnu_Band3",
"Flux_Band1","nuFnu_Band1","LP_beta","LP_SigCurv","Energy_Flux100","PLEC_SigCurv",
"Flux_Band4","nuFnu_Band4","nuFnu_Band5","Flux_Band5","Flux1000","Signif_Avg",
"PLEC_Exp_Index","Source_Name","ASSOC_FGL","CLASS1","ASSOC1"
------"""

features_of_interest = ["PL_Index","LP_Index","Pivot_Energy","PLEC_Flux_Density","LP_Flux_Density","PL_Flux_Density","Frac2_Variability","nuFnu_Band7","PLEC_Index","Flux_Band2","Flux_Band7","nuFnu_Band2","Frac_Variability","PLEC_Expfactor","Variability2_Index","Variability_Index","nuFnu_Band6","Flux_Band3","Npred","Flux_Band6","nuFnu_Band3","Flux_Band1","nuFnu_Band1","LP_beta","LP_SigCurv","Energy_Flux100","PLEC_SigCurv","Flux_Band4","nuFnu_Band4","nuFnu_Band5","Flux_Band5","Flux1000","Signif_Avg","PLEC_Exp_Index","Source_Name","ASSOC_FGL","CLASS1","ASSOC1"]

input_filename = './blazar_catalog_v35.csv'
output_filename = './blazar_catalog_v35_features.csv'
df = pd.read_csv(input_filename)
```

Check if the features exists in the catalog file and extract the ones that exist. Save the output to a file.

```python
valid_features = [col for col in features_of_interest if col in df.columns]
print(f"Number of features of interest = {len(features_of_interest)},\
       Number of features found in the database = {len(valid_features)}")
selected_df = df[valid_features]
selected_df.to_csv(output_filename, index=False)
print(f"Successfully extracted existing features to '{output_filename}'")
```

- Extract the features with scalar values by removing the columns with matrix
- Columns such as "Flux_History" are stored as string in the csv file. Add a literal_eval function to convert them back to Python object.

```python
non_scalar_columns = ['Flux_Band', 'Unc_Flux_Band', 'nuFnu_Band', 'Sqrt_TS_Band', 'Flux_History', 'Unc_Flux_History', 'Sqrt_TS_History']
new_df = df.drop(columns=non_scalar_columns, errors='ignore')

new_df.to_csv("blazar_catalog_v35_single_value_columns.csv", index=False)
print(f"The columns with multiple values {non_scalar_columns}")
```
